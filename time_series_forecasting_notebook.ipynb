## 2.1 Visão Geral: contagem e valores em falta

<details>
<summary><strong> 2.1 Visão Geral - Insights </strong> (Expandir)</summary>

Nesta secção, obtivemos uma visão geral do subconjunto de dados selecionado para a loja CA\_1 do departamento FOODS\_1 (`Y_df_FOODS1_CA1`). Confirmámos os seguintes pontos essenciais para garantir a qualidade dos dados antes da modelagem:

*   Existe um total de **216 séries temporais únicas**.
*   As datas abrangem o período de **29-01-2011 a 19-06-2016**.
*   Não foram encontradas **linhas duplicadas** nem **valores negativos** na coluna de vendas (`y`), o que é fundamental para a integridade dos dados de vendas.
*   Não existem **valores em falta** em nenhuma das colunas relevantes (`unique_id`, `ds`, `y`).
*   Identificámos que uma percentagem significativa das observações (**51.84%**) corresponde a **vendas zero**. Este é um ponto importante a considerar durante a modelagem, pois indica uma elevada intermitência nas vendas de muitos produtos.

Esta análise inicial permitiu validar a consistência dos dados e identificar características importantes como a frequência de vendas zero, que influenciarão as abordagens de previsão.
</details>

```python
# 2.1 Visão Geral: contagem e valores em falta

print("2.1 Visão Geral: \n")

# Calcular o número de séries únicas
n_series = Y_df_FOODS1_CA1['unique_id'].nunique()

# Calcular valores em falta por coluna
missing = Y_df_FOODS1_CA1.isna().sum()

# Encontrar a data mínima e máxima na coluna 'ds'
date_min = Y_df_FOODS1_CA1['ds'].min()
date_max = Y_df_FOODS1_CA1['ds'].max()

# Calcular o número de linhas duplicadas
# O parâmetro keep=False marca todas as duplicadas como True
n_duplicates = Y_df_FOODS1_CA1.duplicated(keep=False).sum()

# Verificar se existem valores negativos na coluna 'y'
negative_sales_count = (Y_df_FOODS1_CA1['y'] < 0).sum()

print(f"Número de séries temporais únicas: {n_series}")
print(f"\nPeríodo das séries temporais: De {date_min.strftime('%d-%m-%Y')} a {date_max.strftime('%d-%m-%Y')}")
print(f"\nNúmero de linhas duplicadas: {n_duplicates}")
print("\nValores em falta por coluna:")
print(missing)
print(f"\nNúmero de valores negativos na coluna 'y': {negative_sales_count}")
```

## 2.2 Análise Agregada de Vendas

<details>
<summary><strong> 2.2 Análise Agregada - Insights </strong> (Expandir)</summary>

Nesta secção, a análise focou-se nas vendas totais do departamento `FOODS_1` na loja `CA_1` para identificar padrões gerais e sazonalidades:

*   O **plot geral das vendas diárias agregadas** mostrou a tendência e o comportamento global das vendas ao longo do tempo.
*   A análise da **sazonalidade semanal** revelou como as vendas médias variam consistentemente ao longo dos dias da semana.
*   A análise da **sazonalidade mensal** identificou padrões na média das vendas em diferentes meses do ano.
*   A **decomposição da série agregada** separou a série temporal em componentes de tendência, sazonalidade e resíduo, permitindo visualizar a estrutura subjacente dos dados de vendas agregadas.
</details>

```python
# 2.2.1 Análise Agregada de Vendas - Plot Geral

print("2.2.1 ANÁLISE AGREGADA DE VENDAS - PLOT GERAL\n")

y_agg = Y_df_FOODS1_CA1.groupby('ds')['y'].agg(['sum','mean','median'])
plt.figure(figsize=(10,4))
y_agg['sum'].plot(title='Vendas Totais Diárias Agregadas')
plt.show()
```

```python
# 2.2.2 Análise Agregada de Vendas - Weekly and Monthly

print("2.2.2 ANÁLISE AGREGADA DE VENDAS - WEEKLY AND MONTHLY\n")

# Total sales by day across all products
daily_sales = Y_df_FOODS1_CA1.groupby('ds')['y'].sum().reset_index()
daily_sales['ds'] = pd.to_datetime(daily_sales['ds'])

fig, axes = plt.subplots(2, 1, figsize=(8, 8))

# Weekly seasonality
daily_sales['day_of_week'] = daily_sales['ds'].dt.dayofweek
weekly_pattern = daily_sales.groupby('day_of_week')['y'].mean()
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
axes[0].bar(range(7), weekly_pattern.values)
axes[0].set_title('Average Sales by Day of Week', fontsize=14)
axes[0].set_xlabel('Day of Week')
axes[0].set_ylabel('Average Sales')
axes[0].set_xticks(range(7))
axes[0].set_xticklabels(days)

# Monthly seasonality
daily_sales['month'] = daily_sales['ds'].dt.month
monthly_pattern = daily_sales.groupby('month')['y'].mean()
axes[1].bar(range(1, 13), monthly_pattern.values)
axes[1].set_title('Average Sales by Month', fontsize=14)
axes[1].set_xlabel('Month')
axes[1].set_ylabel('Average Sales')
axes[1].set_xticks(range(1, 13))

fig.suptitle('Análise Agregada de Vendas\n Departamento: FOODS 1 - Loja: CA 1')

plt.tight_layout()
plt.show()
```

```python
# 2.2.3 Decomposição na série agregada

print("2.2.3 DECOMPOSIÇÃO DA SÉRIE AGREGADA \n")

decomposition = seasonal_decompose(Y_df_FOODS1_CA1.groupby('ds')['y'].sum(), model='additive', period=365)

fig, axes = plt.subplots(4, 1, figsize=(15, 12))
decomposition.observed.plot(ax=axes[0], title='Série Temporal Original')
decomposition.trend.plot(ax=axes[1], title='Componente de Tendência')
decomposition.seasonal.plot(ax=axes[2], title='Componente Sazonal')
decomposition.resid.plot(ax=axes[3], title='Componente Residual')

fig.suptitle('Decomposição Sazonal das Vendas', fontsize=18)

plt.tight_layout()
plt.show()
```

<details>
<summary><strong> 2.3 Análise Individual - Insights </strong> (Expandir)</summary>

Nesta secção, explorámos o comportamento das vendas ao nível de cada produto:

*   **Top Produtos por Volume de Vendas:** Analisámos os produtos com maior volume de vendas, notando a variabilidade nas suas vendas diárias através de gráficos.

```python
# 2.3.1 Análise Individual de Produtos (Séries) - Visão Geral

print("2.3.1 ANÁLISE INDIVISUAL DE PRODUTOS - VISÃO GERAL \n")

# Encontrar séries temporais mais longa e mais curta
print("Análise do comprimento das Séries Temporais Individuais: \n")

# Criar cópia limpa com categorias filtradas
Y_df_FOODS1_CA1_clean = Y_df_FOODS1_CA1.copy()
Y_df_FOODS1_CA1_clean['unique_id'] = Y_df_FOODS1_CA1_clean['unique_id'].cat.remove_unused_categories()

# 1. Agrupar por unique_id e calcular o comprimento de cada série
series_lengths = Y_df_FOODS1_CA1_clean.groupby('unique_id', observed=True).size()

# 2. Encontrar série mais longa e mais curta
max_id = series_lengths.idxmax()
min_id = series_lengths.idxmin()

max_length = series_lengths[max_id]
min_length = series_lengths[min_id]

# 3. Obter as datas para cada uma dessas séries
max_series_dates = Y_df_FOODS1_CA1_clean[Y_df_FOODS1_CA1_clean['unique_id'] == max_id]['ds']
min_series_dates = Y_df_FOODS1_CA1_clean[Y_df_FOODS1_CA1_clean['unique_id'] == min_id]['ds']

# 4. Mostrar os resultados
print(f"Série mais longa: {max_id}")
print(f"Número de observações: {max_length}")
print(f"Intervalo de datas: {max_series_dates.min().strftime('%d-%m-%Y')} até {max_series_dates.max().strftime('%d-%m-%Y')} \n")

print(f"Série mais curta: {min_id}")
print(f"Número de observações: {min_length}")
print(f"Intervalo de datas: {min_series_dates.min().strftime('%d-%m-%Y')} até {min_series_dates.max().strftime('%d-%m-%Y')} \n")

# Calculate zero sales percentage
zero_days = (Y_df_FOODS1_CA1['y'] == 0).sum()
print(f'Dias com venda zero: {zero_days} ({zero_days / len(Y_df_FOODS1_CA1):.2%})')
```

```python
# 2.3.2 Análise Individual de Produtos (Séries) - Volume de Vendas (geral)

print("2.3.2 ANÁLISE INDIVIDUAL DE PRODUTOS - VOLUME DE VENDAS: GERAL\n")

# Statisticas por produto
product_stats = Y_df_FOODS1_CA1_clean.groupby('unique_id', observed=True)['y'].agg([
    'count', 'mean', 'std', 'min', 'max', 'sum'
]).round(2)

# Ordenar os dados por volume total de vendas
product_stats_sorted = product_stats.sort_values(by='sum', ascending=False).reset_index()

# Produto com maior volume de vendas
max_sales = product_stats['sum'].max()
max_product = product_stats['sum'].idxmax()
print(f"Produto com MAIOR volume de vendas: {max_product} ({max_sales:,.0f} vendas) \n")

# Produto com menor volume de vendas
min_sales = product_stats['sum'].min()
min_product = product_stats['sum'].idxmin()
print(f"Produto com MENOR volume de vendas: {min_product} ({min_sales:,.0f} vendas)")

print("\n")

# Gráfico de linha com todos os volumes de vendas
plt.figure(figsize=(14, 6))
sns.lineplot(
    data=product_stats_sorted,
    x=range(len(product_stats_sorted)),
    y='sum',
    marker='o',
    linewidth=1.5,
    color='steelblue'
)

plt.title("Volume Total de Vendas por Série Temporal", fontsize=14)
plt.xlabel("Séries Temporais Ordenadas por Volume de Vendas", fontsize=12)
plt.ylabel("Volume Total de Vendas", fontsize=12)
plt.xticks([], [])  # ocultar labels do eixo x para legibilidade
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# 2.3.3 Análise Individual de Produtos (Séries) - Volume de Vendas (Top dos produtos mais vendidos)

print("2.3.3 ANÁLISE INDIVIDUAL DE PRODUTOS - VOLUME DE VENDAS: TOP PRODUTOS MAIS VENDIDOS\n")

print("Top 10 produtos por total de vendas:")
print(product_stats.sort_values('sum', ascending=False).head(10))

print('\n')

# Visualizar top produtos
top_products = product_stats.sort_values('sum', ascending=False).head(6)['sum']

fig, axes = plt.subplots(3, 2, figsize=(20, 13))
axes = axes.flatten()

for i, product_id in enumerate(top_products.index):
    product_data = Y_df_FOODS1_CA1_clean[Y_df_FOODS1_CA1_clean['unique_id'] == product_id].copy()
    product_data['ds'] = pd.to_datetime(product_data['ds'])

    axes[i].plot(product_data['ds'], product_data['y'], alpha=0.8)
    axes[i].set_title(f'{product_id}\nVendas totais: {top_products[product_id]:.0f}', fontsize=12)
    axes[i].tick_params(axis='x', rotation=45)
    axes[i].set_ylabel('Daily Sales')

fig.suptitle('Análise individual dos Produtos \n Departamento: FOODS 1 - Loja: CA 1', fontsize=18, y=1.02)

plt.tight_layout()
plt.show()
```

## 2.4 Estacionaridade

<details>

Nesta secção, analisámos a estacionaridade das séries temporais para entender a sua estrutura de dependência ao longo do tempo:

*   A análise da **série agregada** das vendas indicou que esta é **estacionária** (p-value do teste ADF de 0.0040, menor que 0.05).
*   Ao nível dos **produtos individuais**, a grande maioria das séries (206 em 216, ou **95.4%**) mostrou ser **estacionária**.
*   Identificámos **10 séries** como **não estacionárias** (4.6%), correspondendo aos produtos com IDs: `019`, `023`, `057`, `082`, `087`, `105`, `115`, `118`, `163` e `170`.
*   Para as séries não estacionárias, foi realizada uma **decomposição** para visualizar as suas componentes de tendência e sazonalidade.
</details>

<details>
<summary><strong> Funções auxiliares </strong> (Expandir)</summary>

<details>
<summary><strong> adf_test() </strong> (Explicação)</summary>

* Realiza o teste de Dickey-Fuller Aumentado (ADF) numa série temporal.
* Retorna um dicionário com a estatística do teste, p-value, valores críticos e se a série é estacionária.
</details>

<details>
<summary><strong> stationarity_analysis() </strong> (Explicação)</summary>

* Analisa a estacionaridade das vendas agregadas e de cada produto individualmente usando o teste ADF.
* Imprime um resumo da estacionaridade (total, estacionárias, não estacionárias) e lista as séries não estacionárias.
* Retorna um DataFrame com os resultados de todos os testes e o resultado agregado.
</details>

<details>
<summary><strong> plot_stationarity_results() </strong> (Explicação)</summary>

* Cria dois gráficos para visualizar os resultados da análise de estacionaridade:
  * Um histograma mostra a distribuição dos p-values do teste ADF.
  * Um gráfico de pizza exibe a proporção de séries estacionárias versus não estacionárias.
</details>

<details>
<summary><strong> plot_example_series() </strong> (Explicação)</summary>

* Plota exemplos de séries temporais para ilustrar a diferença entre séries estacionárias e não estacionárias.
* Seleciona as duas primeiras séries estacionárias e as duas primeiras não estacionárias para visualização.
</details>

</details>

```python
# funções auxiliares

# teste adf

def adf_test(series):
    result = adfuller(series, autolag='AIC')
    return {
        'test_statistic': result[0],
        'p_value': result[1],
        'used_lag': result[2],
        'n_obs': result[3],
        'critical_values': result[4],
        'is_stationary': result[1] <= 0.05
    }

# loop pelas séries

def stationarity_analysis(data):
    print("STATIONARITY ANALYSIS\n")
    results = []

    # 1. AGGREGATE SALES
    aggregate_series = data.groupby('ds')['y'].sum()
    agg_result = adf_test(aggregate_series)

    print("1. AGGREGATE SALES STATIONARITY:")
    print(f"   ADF p-value: {agg_result['p_value']:.4f}")
    print(f"   Is Stationary: {agg_result['is_stationary']}\n")

    # 2. INDIVIDUAL SERIES
    print("2. INDIVIDUAL PRODUCTS:")

    for uid in data['unique_id'].unique():
        serie = data[data['unique_id'] == uid].sort_values(by='ds')['y']
        res = adf_test(serie)
        res['product'] = uid
        results.append(res)

    results_df = pd.DataFrame(results)
    total = len(results_df)
    stationary = results_df['is_stationary'].sum()
    non_stationary = total - stationary

    print(f"   Stationary: {stationary}/{total} ({100*stationary/total:.1f}%)")
    print(f"   Non-stationary: {non_stationary}/{total} ({100*non_stationary/total:.1f}%)\n")

    # Séries não estacionárias
    non_stat_df = results_df[~results_df['is_stationary']][['product', 'p_value', 'used_lag']]
    print("   List of Non-Stationary Series:")
    print(non_stat_df.to_string(index=False))

    return results_df, agg_result

# visualização dos resultados

def plot_stationarity_results(results_df):
    fig, axes = plt.subplots(1, 2, figsize=(15, 5))

    # Histograma de p-valores
    axes[0].hist(results_df['p_value'], bins=20, alpha=0.7, edgecolor='black')
    axes[0].axvline(x=0.05, color='red', linestyle='--', label='α = 0.05')
    axes[0].set_title('ADF Test p-values Distribution')
    axes[0].set_xlabel('p-value')
    axes[0].set_ylabel('Frequency')
    axes[0].legend()

    # Gráfico de pizza
    counts = results_df['is_stationary'].value_counts().sort_index()
    labels = ['Non-Stationary', 'Stationary']
    axes[1].pie(counts.values, labels=labels, autopct='%1.1f%%', colors=['lightcoral', 'lightblue'])
    axes[1].set_title('Stationarity Distribution')

    plt.tight_layout()
    plt.show()

# visualização dos exemplos

def plot_example_series(data, results_df):
    stationary_ids = results_df[results_df['is_stationary']]['product'].head(2)
    non_stationary_ids = results_df[~results_df['is_stationary']]['product'].head(2)

    fig, axes = plt.subplots(2, 2, figsize=(15, 8))

    examples = [(stationary_ids, 'Stationary', 0), (non_stationary_ids, 'Non-Stationary', 1)]

    for ids, label, row in examples:
        for i, uid in enumerate(ids):
            sub = data[data['unique_id'] == uid].sort_values(by='ds')
            sub['ds'] = pd.to_datetime(sub['ds'])

            axes[row, i].plot(sub['ds'], sub['y'])
            axes[row, i].set_title(f'{label} Example {i+1} - {uid}')
            axes[row, i].tick_params(axis='x', rotation=45)

    plt.tight_layout()
    plt.show()
```

### Definir funções auxiliares e constantes

<details>
<summary><strong> Funções Auxiliares e preparação dos modelos </strong> (Expandir)</summary>

Para iniciar os modelos, recorreu-se a funções auxiliares para iniciar, ainda antes de correr os modelos, o armazenamento das suas métricas de forma a permitir uma comparação entre modelos no final:

*   A função `model_evaluation` foi definida para calcular métricas de desempenho comuns (MSE, MAE, RMSE, MAPE) a partir dos valores reais (`y_true`) e das previsões (`y_pred`) de um modelo.
*   Um DataFrame vazio (`Models_results`) foi inicializado para armazenar os resultados da avaliação de todos os modelos implementados, facilitando a comparação posterior.
*   A frequência (`freq`) dos dados foi definida como diária ('D').
*   O horizonte de previsão (`horizon`) foi definido com base no número de dias únicos no conjunto de teste, garantindo que as previsões abranjam todo o período de avaliação.

Com a preparação dos dados e a definição das ferramentas de avaliação concluídas, estamos agora prontas para iniciar a implementação e o treino dos modelos de previsão.

</details>

```python
# função auxiliar de avaliação

def model_evaluation(y_true, y_pred, Model):
    results_row = [{"Model": Model,
                    "MSE": metrics.mean_squared_error(y_true, y_pred),
                    "MAE": metrics.mean_absolute_error(y_true, y_pred),
                    "RMSE": np.sqrt(metrics.mean_squared_error(y_true, y_pred)),
                    "MAPE": metrics.mean_absolute_percentage_error(y_true, y_pred) * 100,
                   }]

    results = pd.DataFrame.from_records(results_row)
    return results
```

```python
# grid de parametros a testar
lgbm_param_grid = {
    'num_leaves': [31, 50],
    'n_estimators': [100, 200],
    'learning_rate': [0.1]
}

# gerar todas as combinações
from itertools import product
lgbm_param_list = list(product(
    lgbm_param_grid['num_leaves'],
    lgbm_param_grid['n_estimators'],
    lgbm_param_grid['learning_rate']
))
```

```python
print("Modelo: LightGBM \n")

# Loop pela grid
lightgbm_results = []

for num_leaves, n_estimators, lr in tqdm(
        lgbm_param_list, desc='LGBM tuning'):

    model_name = f"LGBM_nl{num_leaves}_ne{n_estimators}"

    # instanciar o LightGBM
    lgbm = LGBMRegressor(
        num_leaves=num_leaves,
        n_estimators=n_estimators,
        learning_rate=lr,
        random_state=42
    )

    mlf = MLForecast(
        models=[lgbm],
        freq=freq,
        lags=list(range(1, 7)),
        lag_transforms={
            1: [ExpandingMean()],
            7: [RollingMean(window_size=7)]},
        date_features=['year', 'month', 'day', 'dayofweek', 'quarter', 'week'],
    )

    # treinar no conjunto de treino
    mlf.fit(train)

    # prever para o horizonte completo
    fcst = mlf.predict(horizon)

    # extrair vetores y_pred e y_true
    y_pred = fcst['LGBMRegressor'].values
    y_true = test.sort_values(['unique_id', 'ds'])['y'].values

    # Avaliar e guardar como DataFrame
    evaluation_metrics = model_evaluation(y_true, y_pred, model_name)
    lightgbm_results.append(evaluation_metrics)
```

```python
print('Modelo: NBEATS')

# grid de parâmetros para NBEATS
nbeats_param_grid = {
    'input_size': [14, 28],
    'h': [horizon],
    'max_steps': [100, 200],
    'stack_types': [['identity', 'trend', 'seasonality']]
}

# gerar todas as combinações
from itertools import product
nbeats_param_list = list(product(
    nbeats_param_grid['input_size'],
    nbeats_param_grid['h'],
    nbeats_param_grid['max_steps'],
    nbeats_param_grid['stack_types']
))

print("Grid Search para NBEATS")

# loop pela grid
nbeats_results = []

for input_size, h, max_steps, stack_types in tqdm(
        nbeats_param_list, desc='NBEATS tuning'):

    model_name = f"NBEATS_is{input_size}_ms{max_steps}"

    # instanciar o NBEATS
    nbeats_model = NBEATS(
        h=h,
        input_size=input_size,
        max_steps=max_steps,
        stack_types=stack_types,
        loss=MAE(),
        start_padding_enabled=True
    )

    # criar NeuralForecast
    nf = NeuralForecast(
        models=[nbeats_model],
        freq=freq
    )

    try:
        # treinar no conjunto de treino
        nf.fit(train)

        # fazer previsões
        fcst = nf.predict()

        # etrair vetores y_pred e y_true
        y_pred = fcst['NBEATS'].values
        y_true = test.sort_values(['unique_id', 'ds'])['y'].values

        # avaliar e guardar como DataFrame
        evaluation_metrics = model_evaluation(y_true, y_pred, model_name)
        nbeats_results.append(evaluation_metrics)

    except Exception as e:
        print(f"Erro no modelo {model_name}: {str(e)}")
        continue
```

```python
print("Modelo: NHITS \n")

# grid de parâmetros para NHITS
nhits_param_grid = {
    'input_size': [14, 28],
    'h': [horizon],
    'max_steps': [100, 200],
    'n_pool_kernel_size': [[2, 2, 1], [4, 2, 1]]
}

# gerar todas as combinações
nhits_param_list = list(product(
    nhits_param_grid['input_size'],
    nhits_param_grid['h'],
    nhits_param_grid['max_steps'],
    nhits_param_grid['n_pool_kernel_size']
))
```

```python
print("Grid Search para NHITS iniciado...")

# loop pela grid
nhits_results = []

for input_size, h, max_steps, n_pool_kernel_size in tqdm(
        nhits_param_list, desc='NHITS tuning'):

    model_name = f"NHITS_is{input_size}_ms{max_steps}_pk{len(n_pool_kernel_size)}"

    # instanciar o NHITS
    nhits_model = NHITS(
        h=h,
        input_size=input_size,
        max_steps=max_steps,
        n_pool_kernel_size=n_pool_kernel_size,
        loss=MAE()
    )

    # criar NeuralForecast
    nf_nhits = NeuralForecast(
        models=[nhits_model],
        freq=freq
    )

    try:
        # treinar conjunto de treino
        nf_nhits.fit(train)

        # fazer previsões
        fcst = nf_nhits.predict()

        # extrair vetores y_pred e y_true
        y_pred = fcst['NHITS'].values
        y_true = test.sort_values(['unique_id', 'ds'])['y'].values

        # avaliar e guardar DataFrame
        evaluation_metrics = model_evaluation(y_true, y_pred, model_name)
        nhits_results.append(evaluation_metrics)

    except Exception as e:
        print(f"Erro no modelo {model_name}: {str(e)}")
        continue
```

### Análise de Cobertura dos Intervalos de Confiança

<details>
<summary><strong> Cobertura observada </strong> (Expandir)</summary>

Com o modelo `NBEATS com regressores` identificado como o de melhor desempenho, gerámos previsões pontuais e intervalos de confiança para o período de teste.

- **Cobertura observada 90% (área mais clara/externa):** 93,98%  
  De igual modo, o intervalo de 90% cobriu quase 94% dos valores reais, resultando numa sobrecobertura de cerca de 4 pontos percentuais. - Tamém é possível ver a **previsão pontual (linha sólida no gráfico)** calculada através da mediana.
</details>

<details>
<summary><strong> Interpretação da Sobrecobertura </strong> (Expandir)</summary>

- Estes intervalos indicam a gama de valores dentro da qual esperamos que as vendas reais caiam com uma certa probabilidade (80% ou 90%). A largura do intervalo reflete a incerteza da previsão do modelo; intervalos mais largos indicam maior incerteza.

- Um intervalo nominal de `X`% deveria cobrir aproximadamente `X`% dos pontos observados. Aqui, os intervalos estão próximos, mas com tendência a serem conservadores:  
  - Ao abranger 85% em vez de 80%, ficamos “mais seguros”, mas sacrifica‐se alguma precisão/*sharpness*.  
  - Os limites poderiam ser estreitados sem perder a cobertura nominal.

**Prós** de ter Sobrecobertura  
- Menor probabilidade de *errar por fora*: a previsão fica mais cautelosa, reduzindo o risco de o valor real ficar fora do intervalo.  
- Maior robustez em cenários críticos (e.g., gestão de stocks, cálculo de coberturas de risco), onde falhar por fora pode ser dispendioso.

**Contras** de ter Sobrecobertura  
- Limites mais largos do que o necessário podem conduzir a decisões excessivamente conservadoras.  
- Em situações em que se exige maior precisão (e.g., planeamento financeiro, gestão de inventário), intervalos demasiado amplos implicam perda de potencial informativo.
</details>

<details>
<summary><strong> Sugestões para Ajustar a Cobertura </strong> (Expandir)</summary>

Se for necessário reduzir a sobrecobertura e aproximar a cobertura observada do valor nominal (e.g., 80% exato), pode-se: retreinar o modelo, ajustar os pesos dos quantis ou testar outras distribuição.
</details>

<details>
<summary><strong> Conclusões </strong> (Expandir)</summary>

**Conclusão:** Na previsão há sempre incerteza e os intervalos de confiança quantificam essa margem de erro, servindo de suporte à decisão. Apesar de ligeiramente conservadores, os intervalos apresentam boa cobertura e oferecem margem de segurança para a tomada de decisões.

Consequentemente, o erro pontual da mediana acaba por ser um pouco maior do que num modelo treinado exclusivamente para minimizar o MAE.

Em suma, não se trata de uma diferença aleatória, mas sim de um efeito direto do novo objetivo do treino: sacrificar um pouco a precisão pontual em favor de gerar intervalos de confiança bem calibrados.

##  Conclusão

Neste trabalho, aplicámos e comparamos diversos modelos de previsão de séries temporais no departamento **FOODS_1 da loja CA_1**, utilizando o conjunto de dados do **M5 Forecasting**. Através de uma análise exploratória detalhada e da avaliação do desempenho dos modelos com métricas como MAE, MSE, RMSE e MAPE, destacámos o **NBEATS com regressores** como o modelo com melhor desempenho geral.

A análise exploratória de dados (EDA) revelou uma característica fundamental do problema: cerca de 51,84% dos registos de vendas são zeros, evidenciando um padrão de demanda intermitente. Esta observação tem implicações diretas na escolha e interpretação das métricas de avaliação. Em particular, a métrica MAPE (erro percentual absoluto médio) mostrou-se pouco fiável, já que valores reais nulos ou muito baixos amplificam desproporcionalmente os erros percentuais, conduzindo a resultados inflacionados e por vezes enganadores. Por esta razão, a interpretação dos resultados deu prioridade a métricas mais robustas, como MAE, MSE e RMSE, que fornecem uma visão mais estável do desempenho dos modelos.

O modelo **NBEATS com regressores** mostrou-se particularmente eficaz a capturar padrões complexos da série temporal, fornecendo previsões com o menor erro absoluto médio (MAE). Além disso, a capacidade do NBEATS de gerar previsões probabilísticas — através de intervalos de confiança — fornece uma ferramenta essencial para a gestão de incerteza e para apoiar decisões operacionais em ambientes voláteis.

A **inclusão de regressores** demonstrou um impacto positivo no desempenho de alguns modelos, como o XGBoost e o NBEATS. Contudo, é importante salientar que os modelos foram inicialmente ajustados **sem a presença desses regressores** e, posteriormente, os mesmos hiperparâmetros foram reutilizados com os regressores adicionados. Este procedimento, embora útil para efeitos comparativos diretos, não permite avaliar plenamente o contributo dos regressores. Idealmente, o processo de ajuste de **hiperparâmetros (fine tuning)** deveria ter sido repetido após a inclusão dos regressores, uma vez que a sua presença altera a estrutura do espaço de procura e pode influenciar significativamente a configuração ótima do modelo. Assim, é provável que os benefícios observados dos regressores estejam subestimados, e que uma nova otimização dedicada revelasse ganhos adicionais relevantes.

No caso do **AutoARIMA**, observou-se um comportamento inverso, a versão **sem regressores** teve ligeiramente melhor desempenho, o que pode indicar que a inclusão de variáveis exógenas sem uma reconfiguração adequada do modelo não traz vantagens, e pode até pode introduzir ruído.

Por fim, os modelos de **Deep Learning (NBEATS e AutoLSTM)** apresentaram MAPE's mais baixos do que os modelos estatísticos e baseados em árvores, sugerindo **maior precisão relativa** em termos percentuais — ainda que, por vezes, os erros absolutos (MAE, RMSE) fossem ligeiramente superiores aos do AutoARIMA. Esta diferença evidencia que diferentes métricas capturam diferentes aspetos do erro, e reforça a necessidade de uma avaliação multidimensional do desempenho.

##Principais insights do estudo:

*   A elevada proporção de vendas nulas compromete o MAPE como métrica de avaliação, sendo preferível usar MAE, MSE e RMSE.
*   O NBEATS com regressores foi o modelo mais eficaz, fornecendo previsões pontuais e probabilísticas com elevada precisão.
*   A inclusão de regressores melhorou o desempenho de alguns modelos, mas o seu impacto real foi provavelmente subestimado devido à ausência de nova afinação.
*   A repetição do fine tuning com os regressores incluídos desde o início é fundamental para explorar todo o seu potencial.
*   Modelos de Deep Learning capturam melhor padrões não lineares e apresentaram bom desempenho mesmo com dados ruidosos.


## Sugestões de Melhoria
Para futuras iterações e otimização dos modelos de previsão, sugerimos as seguintes melhorias:
* **Introdução de Intervalos de Previsão para todos os modelos:** Além das previsões pontuais, é fundamental fornecer intervalos de previsão (probabilistic forecasts). Estes intervalos oferecem uma medida da incerteza associada às previsões, permitindo uma melhor gestão de risco e planeamento.
* **Otimização com Regressores:** Realizar uma nova ronda de afinação (tuning) dos modelos que utilizam regressores, garantindo que os regressores são considerados desde o início do processo de otimização. Isso pode desbloquear um desempenho superior que não foi alcançado na presente análise devido a limitações de recursos.
* **Identificação do Melhor Modelo por Série Individual:** Em vez de procurar um "melhor modelo geral", pode ser mais eficaz identificar o melhor modelo para cada série temporal individual. Para isso, recomenda-se o treino paralelo de múltiplos modelos sobre cada série e a posterior seleção do mais adequado com base em métricas de erro relevantes (como MAE, RMSE, etc.). Desta forma, diferentes padrões de dados podem ser mais bem capturados por diferentes algoritmos, levando a previsões mais precisas em um cenário desagregado.
* **Exploração de Outros Regressores:** Investigar e testar a inclusão de outros regressores potencialmente relevantes que possam capturar melhor a dinâmica dos dados (e.g., feriados, eventos especiais, indicadores económicos, etc.).
* **Ensemble Modeling:** Considerar a criação de modelos de ensemble, combinando as previsões de vários modelos para potencialmente reduzir o erro e aumentar a robustez das previsões.
